# EDA: Viable Movies Universe

Exploratory analysis of the data in PostgreSQL for the movie-success-prediction project.

**Universe filter.** Only movies that have BOTH:
- `opening_weekend_revenue` from The Numbers (`box_office` table), AND
- a non-null budget — `box_office.production_budget` when present, else TMDB's `movies_2025.budget` (via `COALESCE`).

This is the same filter applied to the trailer-comments pipeline.

**Figures.** Saved as PDF under `python/notebooks/figures/`. Drop into Overleaf with `\includegraphics{figures/<name>.pdf}` (vector, scales perfectly).

**Dependencies.** Requires `jupyterlab`, `ipykernel`, `matplotlib`, `seaborn` (added to `requirements.txt`). The rest (`pandas`, `sqlalchemy`, `psycopg2-binary`) are already there.

## 0. Setup

````python
import os
import sys
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sqlalchemy import create_engine

# Find python/ regardless of where the notebook is launched from, then chdir
# into it so src.config can resolve ./config.yaml.
def _find_python_root() -> Path:
    cur = Path().resolve()
    for parent in [cur] + list(cur.parents):
        if (parent / "src" / "config.py").exists():
            return parent
    raise RuntimeError("Could not locate python/src/config.py from this notebook")

PYTHON_ROOT = _find_python_root()
os.chdir(PYTHON_ROOT)
sys.path.insert(0, str(PYTHON_ROOT))

from src import config  # noqa: E402

engine = create_engine(
    f"postgresql+psycopg2://{config.POSTGRE_USER}:{config.POSTGRE_PASS}"
    f"@{config.POSTGRE_HOST}:{config.POSTGRE_PORT}/{config.DB_NAME}"
)

FIGURES_DIR = PYTHON_ROOT / "notebooks" / "figures"
FIGURES_DIR.mkdir(parents=True, exist_ok=True)

sns.set_theme(context="paper", style="whitegrid")
plt.rcParams.update({
    "savefig.bbox": "tight",
    "figure.figsize": (8, 5),
    "font.size": 11,
    "axes.titlesize": 12,
    "axes.labelsize": 11,
})

def savefig(name: str, fig=None) -> None:
    """Save the current (or given) figure as PDF under notebooks/figures/."""
    if fig is None:
        fig = plt.gcf()
    out = FIGURES_DIR / f"{name}.pdf"
    fig.savefig(out)
    print(f"Saved {out.relative_to(PYTHON_ROOT)}")

print(f"Connected to {config.DB_NAME}@{config.POSTGRE_HOST}:{config.POSTGRE_PORT}")
print(f"Figures dir: {FIGURES_DIR.relative_to(PYTHON_ROOT)}/")
````

## 1. Load the viable universe

One DataFrame `movies` (joined with `box_office`) and one DataFrame `comments` (also filtered to viable movies).

````python
movies_sql = """
SELECT
    m.*,
    bo.opening_weekend_revenue,
    bo.production_budget,
    bo.opening_theaters,
    bo.the_numbers_url,
    COALESCE(bo.production_budget, m.budget) AS effective_budget
FROM movies_2025 AS m
INNER JOIN box_office AS bo ON m.tmdb_id = bo.tmdb_id
WHERE bo.opening_weekend_revenue IS NOT NULL
  AND COALESCE(bo.production_budget, m.budget) IS NOT NULL
"""
movies = pd.read_sql(movies_sql, engine)
print(f"Viable movies: {len(movies):,}")
movies.head(3)
````

````python
comments_sql = """
SELECT tc.*
FROM trailer_comments AS tc
INNER JOIN movies_2025 AS m  ON tc.tmdb_id = m.tmdb_id
INNER JOIN box_office     AS bo ON tc.tmdb_id = bo.tmdb_id
WHERE bo.opening_weekend_revenue IS NOT NULL
  AND COALESCE(bo.production_budget, m.budget) IS NOT NULL
"""
comments = pd.read_sql(comments_sql, engine)
print(f"Comment rows: {len(comments):,}")
print(f"Movies with at least one comment: {comments['tmdb_id'].nunique():,} of {len(movies):,}")
comments.head(3)
````

## 2. Schema and data types

````python
schema = pd.DataFrame({
    "dtype": movies.dtypes.astype(str),
    "non_null": movies.notna().sum(),
    "null_pct": (movies.isna().mean() * 100).round(2),
    "n_unique": movies.nunique(),
})
schema = schema.sort_values("null_pct", ascending=False)
schema
````

### 4.1 Money columns (log scale)

## 6. Release date over time

## 7. Box-office: budget agreement (TMDB vs The Numbers)

Sanity check: when both sources report a budget, do they agree? Big disagreements suggest data issues.

````python
both = movies.dropna(subset=["production_budget", "budget"]).copy()
both = both[(both["production_budget"] > 0) & (both["budget"] > 0)]
print(f"Movies with BOTH budgets > 0: {len(both):,} ({len(both) / len(movies) * 100:.1f}% of viable)")

fig, ax = plt.subplots(figsize=(6, 6))
ax.scatter(both["budget"], both["production_budget"], s=15, alpha=0.5, color="#4C72B0")
lo, hi = 1e4, max(both["budget"].max(), both["production_budget"].max()) * 1.2
ax.plot([lo, hi], [lo, hi], "k--", lw=1, label="Equal")
ax.set_xscale("log"); ax.set_yscale("log")
ax.set_xlim(lo, hi); ax.set_ylim(lo, hi)
ax.set_xlabel("TMDB budget (USD, log)")
ax.set_ylabel("The Numbers production_budget (USD, log)")
ax.set_title("Budget agreement: TMDB vs The Numbers")
ax.legend()
savefig("budget_agreement")
plt.show()

# Quantify disagreement
both["log_ratio"] = np.log10(both["production_budget"] / both["budget"])
print(f"\nlog10(Numbers/TMDB) — describe:")
print(both["log_ratio"].describe().round(3))
````

````python
# Days BEFORE release the comment was posted (positive = before release).
c = comments[["tmdb_id", "published_at"]].copy()
c["published_at"] = pd.to_datetime(c["published_at"], errors="coerce", utc=True)
c = c.merge(movies[["tmdb_id", "release_date"]], on="tmdb_id", how="left")
c["release_date"] = pd.to_datetime(c["release_date"], errors="coerce", utc=True)
c["days_before_release"] = (c["release_date"] - c["published_at"]).dt.days

days = c["days_before_release"].dropna()
print("Days before release — describe:")
print(days.describe().round(1))

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].hist(days[(days >= 0) & (days <= 365)], bins=80, color="#C44E52")
axes[0].set_xlabel("Days BEFORE release")
axes[0].set_ylabel("Number of comments")
axes[0].set_title("Comments by days-before-release (0-365)")

axes[1].hist(days[(days >= 0) & (days <= 30)], bins=30, color="#C44E52")
axes[1].set_xlabel("Days BEFORE release (zoomed)")
axes[1].set_ylabel("Number of comments")
axes[1].set_title("Final 30 days before release")
plt.tight_layout()
savefig("comments_days_before_release")
plt.show()
````

## 9. Cross-table relationships and target variables

Main modelling target: `opening_weekend_revenue`. But raw dollars are heavy-tailed, and a $50M opening is great for a $10M film and terrible for a $200M film. Two practical alternative targets:

- **`opening_to_budget`** = `opening_weekend_revenue / effective_budget` (regression on a normalized scale).
- **`is_hit`** = `opening_weekend_revenue >= effective_budget` (binary classification — standard industry rough rule).

````python
m = movies.dropna(subset=["opening_weekend_revenue", "effective_budget"]).copy()
m = m[(m["opening_weekend_revenue"] > 0) & (m["effective_budget"] > 0)]

m["opening_to_budget"] = m["opening_weekend_revenue"] / m["effective_budget"]
m["is_hit"] = (m["opening_weekend_revenue"] >= m["effective_budget"]).astype(int)

print(f"Modeling-ready rows: {len(m):,}")
print(f"is_hit rate: {m['is_hit'].mean() * 100:.1f}%")
m[["opening_to_budget", "is_hit"]].describe().round(3)
````

````python
fig, ax = plt.subplots(figsize=(7, 6))
ax.scatter(m["effective_budget"], m["opening_weekend_revenue"],
           s=15, alpha=0.5, color="#4C72B0")
lo, hi = 1e3, m[["effective_budget", "opening_weekend_revenue"]].max().max() * 1.2
ax.plot([lo, hi], [lo, hi], "k--", lw=1, label="opening = budget (break-even-ish)")
ax.set_xscale("log"); ax.set_yscale("log")
ax.set_xlim(lo, hi); ax.set_ylim(lo, hi)
ax.set_xlabel("Effective budget (USD, log)")
ax.set_ylabel("Opening weekend revenue (USD, log)")
ax.set_title("Opening weekend vs. budget")
ax.legend()
savefig("opening_vs_budget")
plt.show()
````

````python
fig, axes = plt.subplots(1, 2, figsize=(11, 4))
axes[0].hist(np.log10(m["opening_to_budget"]), bins=40, color="#4C72B0")
axes[0].axvline(0, color="k", linestyle="--", lw=1, label="opening = budget")
axes[0].set_xlabel("log10(opening_weekend / effective_budget)")
axes[0].set_ylabel("Number of movies")
axes[0].set_title("Opening-to-budget ratio (log10)")
axes[0].legend()

hit_counts = m["is_hit"].value_counts().sort_index()
axes[1].bar(["flop (0)", "hit (1)"], hit_counts.values, color=["#C44E52", "#55A868"])
axes[1].set_ylabel("Number of movies")
axes[1].set_title(f"Binary success label  (hit rate = {m['is_hit'].mean()*100:.1f}%)")
plt.tight_layout()
savefig("target_variables")
plt.show()
````

````python
# Bring n_comments in as a feature
m_corr = movies[["tmdb_id"] + NUMERIC_COLS].merge(
    counts_per_movie, on="tmdb_id", how="left"
)
m_corr["n_comments"] = m_corr["n_comments"].fillna(0)
m_corr = m_corr.drop(columns=["tmdb_id"])

# Spearman because the money columns are heavily skewed
corr = m_corr.corr(method="spearman", numeric_only=True)

fig, ax = plt.subplots(figsize=(9, 8))
sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm", center=0,
            vmin=-1, vmax=1, ax=ax, square=True, cbar_kws={"shrink": 0.8})
ax.set_title("Spearman correlation — numerics + n_comments")
plt.tight_layout()
savefig("correlation_matrix")
plt.show()
````

````python
# Comments-per-movie vs opening weekend — does YouTube buzz track success?
plot_df = movies[["tmdb_id", "opening_weekend_revenue"]].merge(
    counts_per_movie, on="tmdb_id", how="left"
)
plot_df["n_comments"] = plot_df["n_comments"].fillna(0)
plot_df = plot_df[plot_df["opening_weekend_revenue"] > 0]

fig, ax = plt.subplots(figsize=(7, 6))
ax.scatter(plot_df["n_comments"].clip(lower=1),
           plot_df["opening_weekend_revenue"],
           s=15, alpha=0.5, color="#8172B2")
ax.set_xscale("symlog"); ax.set_yscale("log")
ax.set_xlabel("Comments per movie (symlog — includes 0)")
ax.set_ylabel("Opening weekend revenue (USD, log)")
ax.set_title("YouTube comment volume vs. opening-weekend revenue")
savefig("comments_vs_opening")
plt.show()

rho = plot_df["n_comments"].corr(plot_df["opening_weekend_revenue"], method="spearman")
print(f"Spearman ρ (n_comments, opening_weekend_revenue) = {rho:.3f}")
````

## 10. Notes for downstream modeling

- **Targets.** `opening_weekend_revenue` (regression, use log), `opening_to_budget` (regression, use log10), or `is_hit` (classification — note the class balance reported above).
- **Feature transforms.** All money columns should be log-transformed before linear modeling; `vote_count` and `popularity` likewise.
- **Budget source.** Use `effective_budget = COALESCE(production_budget, budget)`. The agreement plot above shows whether the two sources are interchangeable for the rows where both exist.
- **Comment-derived features (next step).** Counts per movie, mean text length, mean like_count, % of comments in the final week before release, sentiment scores — most of these can be aggregated per `tmdb_id` and joined back as features.
- **Selection bias.** The viable filter is implicit conditioning: results generalize to films with measurable theatrical box office, not to streaming-direct / festival releases.
